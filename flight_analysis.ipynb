```python
import pandas as pd
```

# **Loading the file in dataframe**

```python
df=pd.read_csv('/content/Day10_Flight_Operations_Dataset.csv')
```

# **Understanding the dataset**

```python
print("Shape : ",df.shape)
print("\n\nBasic Infomation :")
df.info()
print("\n\nColumns:")
print(df.columns.tolist())
print("\n\nStatistical summary")
df.describe(include='all').T
```

# **Checking Data Quality**

```python
print("Missing values:",df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
```

# **Data Transformation**

```python
df['Flight_Date'] = pd.to_datetime(df['Flight_Date'])
df['Month'] = df['Flight_Date'].dt.month
df['Day_Name'] = df['Flight_Date'].dt.day_name()
df['Estimated_Revenue'] = df['Passengers'] * df['Average_Ticket_Price']
df.head()
```

# **Selection and Filtering**

```python
# Select important columns
important_columns = df[
    ['Flight_ID', 'Airline', 'Origin', 'Destination',
     'Passengers', 'Delay_Minutes', 'Flight_Status']
]
important_columns.head()
```

```python
#Flights delayed by more than 30 minutes
high_delay = df[df['Delay_Minutes'] > 30]
print("Flights delayed by more than 30 minutes:", len(high_delay))
high_delay[['Flight_ID', 'Airline', 'Delay_Minutes', 'Weather']].head()
```

# **Sorting Records**

```python
# Sort by highest delay
df.sort_values(by='Delay_Minutes', ascending=False)[
    ['Flight_ID', 'Airline', 'Delay_Minutes', 'Weather']
].head(10)
```

# **Grouping and Aggregation**

```python
# Performance by airline
airline_analysis = df.groupby('Airline').agg(
    Total_Flights=('Flight_ID', 'count'),
    Total_Passengers=('Passengers', 'sum'),
    Average_Delay=('Delay_Minutes', 'mean'),
    Total_Revenue=('Estimated_Revenue', 'sum'),
    Average_Satisfaction=('Passenger_Satisfaction', 'mean')
).sort_values(by='Total_Revenue', ascending=False)

airline_analysis
```

# **Meaningful Questions and Answers**

```python
# Q1. Which airline generated the highest estimated revenue?
print(airline_analysis['Total_Revenue'].idxmax())
print(airline_analysis['Total_Revenue'].max())
```

```python
# Q2. Which airline has the lowest average delay?
print(airline_analysis['Average_Delay'].idxmin())
print(airline_analysis['Average_Delay'].min())
```

```python
# Q7. What percentage of flights were on time?
on_time= ((df['Flight_Status'] == 'On Time').mean() * 100)
print(f"On-time flights: {on_time:.2f}%")
```

# **Key Observations**

Based on the analysis:

1. **Vistara generated the highest total estimated revenue**, indicating strong overall financial performance in this dataset.
2. **Air India generated the lowest total estimated revenue**
3. **Air India Express had the lowest average delay**
4. **IndiGo had a relatively higher average delay** among the airlines.
5. **On-time flights were the dominant flight status**, accounting for approximately **63% of all flights** in the dataset.
